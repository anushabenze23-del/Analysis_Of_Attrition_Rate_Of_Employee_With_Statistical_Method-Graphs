Company: Green Destinations
Problem: Increase in employee attrition (people leaving the company)

Goal:

1. Find the attrition rate (percentage of employees who left)


2. Find if age, income, or years at company impact attrition

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sn
import plotly.express as px
```

```python

df = pd.read_csv('greendestination.csv')
print(df.head())
```

```python
df.dropna(inplace=True)
df.fillna(0, inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```

```python


left = df[df['Attrition'] == 'Yes'].shape[0]
total = df.shape[0]
attrition_rate = (left / total) * 100
print(f"Attrition Rate: {attrition_rate:.2f}%")
```

```python
attrition_count = df[df['Attrition'] == 'Yes'].shape[0]

print(f"Number of employees who left: {attrition_count}")
```

```python
active_count = df[df['Attrition'] == 'No'].shape[0]

print(f"Number of active employees: {active_count}")
```

```python
# Plot the count of employees by attrition status
sn.countplot(x='Attrition', data=df, palette=['green', 'red'])
plt.title("Employee Attrition Count")
plt.xlabel("Attrition Status")
plt.ylabel("Number of Employees")
plt.show()
```

```python
df[df["Age"] > 30]
```

```python
df["Income_per_Age"] = df["MonthlyIncome"] / df["Age"]
```

```python
df.groupby("Gender")["MonthlyIncome"].mean()
```

```python
attrition_by_age = df[df['Attrition'] == 'Yes'].groupby('Age').size()
plt.figure(figsize=(12, 6))
attrition_by_age.plot(kind='line', marker='o', color='blue')

plt.title("Attrition Trend by Age")
plt.xlabel("Age")
plt.ylabel("Number of Employees Left")
plt.grid(True)
plt.show()
```

```python
left_df = df[df['Attrition'] == 'Yes']
attrition_income = left_df.groupby('MonthlyIncome').size()
attrition_income.plot(kind='line', marker='o', color='orange')

plt.title("Attrition Count by Monthly Income")
plt.xlabel("Monthly Income")
plt.ylabel("Number of Employees Left")
plt.grid(True)
plt.show()
```

```python

left_df = df[df['Attrition'] == 'Yes']
attrition_by_years = left_df['YearsAtCompany'].value_counts().sort_index()
attrition_by_years.plot(kind='bar', color='pink')
plt.title("Attrition Count by Years at Company")
plt.xlabel("Years at Company")
plt.ylabel("Number of Employees Left")
plt.xticks(rotation=0)
plt.show()
```

```python

left_df = df[df['Attrition'] == 'Yes']
attrition_by_gender = left_df['Gender'].value_counts()
attrition_by_gender.plot(kind='bar', color='purple')

plt.title("Attrition Count by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Employees Left")
plt.xticks(rotation=0)
plt.show()
```

```python
active_df = df[df['Attrition'] == 'No']
# Group by Department and count active employees
active_count = active_df['Department'].value_counts()
active_count.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90,
    colors=['#66c2a5', '#fc8d62', '#8da0cb']
)
plt.title("Active Employees by Department")
plt.ylabel("")  # Hide y-axis label
plt.show()

total_per_dept = df['Department'].value_counts()
left_per_dept = df[df['Attrition'] == 'Yes']['Department'].value_counts()
attrition_rate = (left_per_dept / total_per_dept * 100).round(2).fillna(0)
print("\nAttrition Rate by Department:")
print(attrition_rate)
```

```python

left_df = df[df['Attrition'] == 'Yes']
attrition_by_promo = left_df['YearsSinceLastPromotion'].value_counts().sort_index()
attrition_by_promo.plot(kind='bar', color='salmon')
plt.title("Attrition Count by Years Since Last Promotion")
plt.xlabel("Years Since Last Promotion")
```

```python
left_df = df[df['Attrition'] == 'Yes']
attrition_by_mgr = left_df['YearsWithCurrManager'].value_counts().sort_index()
attrition_by_mgr.plot(kind='bar', color='teal')
plt.title("Attrition Count by Years With Current Manager")
plt.xlabel("Years With Current Manager")
plt.ylabel("Number of Employees Left")
plt.xticks(rotation=0)
plt.show()
```

```python

education_field_counts = df['EducationField'].value_counts()
fig = px.pie(
    education_field_counts,
    values=education_field_counts.values,
    names=education_field_counts.index,
    hole=0.6,
    title="Distribution of Employees by Education Field"
)
fig.show()
```

```python

total_count = df['OverTime'].value_counts()
left_count = df[df['Attrition'] == 'Yes']['OverTime'].value_counts()
attrition_rate = (left_count / total_count * 100).round(2)
attrition_rate.plot(kind='bar', color='maroon')
plt.title("Attrition Rate by OverTime")
plt.xlabel("OverTime")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.ylim(0, 100)
plt.show()
```

```python
left_df = df[df['Attrition'] == 'Yes']
attrition_by_satisfaction = left_df['JobSatisfaction'].value_counts().sort_index()
attrition_by_satisfaction.plot(kind='bar', color='skyblue')

plt.title("Attrition Count by Job Satisfaction Level")
plt.xlabel("Job Satisfaction (1 = Low, 4 = High)")
plt.ylabel("Number of Employees Left")
plt.xticks(rotation=0)
plt.show()
```

```python
left = df[df['Attrition'] == 'Yes']['NumCompaniesWorked']
stayed = df[df['Attrition'] == 'No']['NumCompaniesWorked']
plt.hist([left, stayed], bins=range(0, df['NumCompaniesWorked'].max()+2),
         label=['Left', 'Stayed'], color=['red', 'green'], alpha=0.7, edgecolor='black')

plt.title("Histogram: Num Companies Worked vs Attrition")
plt.xlabel("Number of Companies Worked")
plt.ylabel("Number of Employees")
plt.legend()
plt.xticks(range(0, df['NumCompaniesWorked'].max()+1))
plt.show()
```

```python
left_env = df[df['Attrition'] == 'Yes'].groupby('EnvironmentSatisfaction').size()
left_rel = df[df['Attrition'] == 'Yes'].groupby('RelationshipSatisfaction').size()

plt.plot(left_env.index, left_env.values, marker='o', label='Environment Satisfaction', color='blue')
plt.plot(left_rel.index, left_rel.values, marker='o', label='Relationship Satisfaction', color='orange')

plt.title("Attrition Count by Satisfaction Levels")
plt.xlabel("Satisfaction Level (1 = Low, 4 = High)")
plt.ylabel("Number of Employees Left")
plt.legend()
plt.grid(True)
plt.xticks([1, 2, 3, 4])
plt.show()
```

```python
best_employees = df[
    (df['PerformanceRating'] >= 4) &
    (df['JobSatisfaction'] >= 3) &
    (df['EnvironmentSatisfaction'] >= 3) &
    (df['YearsAtCompany'] >= 3) &
    (df['Attrition'] == 'No')
]
print(best_employees[['EmployeeNumber', 'Department', 'JobRole', 'PerformanceRating']])
top = best_employees.sort_values(by=['PerformanceRating', 'YearsAtCompany'], ascending=False)
print(top.head(10))
```

Based on the analysis main causes of attrition:

OverTime:The analysis showed a significantly higher attrition rate for employees who work overtime. This suggests that working extra hours is a strong factor contributing to employees leaving.

Job Satisfaction: The plot of attrition count by job satisfaction level indicates that employees with lower job satisfaction are more likely to leave.

Years with Current Manager and Years Since Last Promotion:The bar plots show that a considerable number of employees who leave have spent a shorter time with their current manager and have not been promoted recently. This could point to issues with management relationships or career growth opportunities.

Years at Company:The attrition count by years at company shows a higher number of employees leaving in the early years of their tenure, particularly in the first 1 to 5 years. This suggests that the initial period at the company might be critical for retention.

Monthly Income:The histogram of monthly income for employees who left and the attrition rate by income bin suggest that lower income levels might be associated with a higher likelihood of attrition
